## Interpretation

The LSTM achieved F1=0.465, ROC-AUC=0.802 with only 5 epochs and stride=12 on CPU.
Performance is substantially below the classical baselines (RF F1=0.978, XGBoost F1=0.978).

**Key reasons for the gap:**
- Only 5 epochs — the model has not converged (val F1 was still 0.28 at best epoch 1)
- stride=12 reduces training windows from 69,937 to 5,829 — 12x less data
- No bidirectional processing, single layer — minimal capacity
- Early stopping triggered at epoch 4 due to patience=3

This result is expected and honest. The LSTM baseline here represents a severely
resource-constrained run. The Transformer (Phase 5) will use the same constraints
for a fair comparison, and the academic report will document these hardware limitations.
